```python
import numpy as np
import pandas as pd
from openai import OpenAI
from scipy.spatial import distance
from ast import literal_eval
```

```python
client = OpenAI(api_key=" ")
```

```python
def get_embedding(text, model="text-embedding-ada-002"):
    edited_text = text.replace("\n", " ")   
    return client.embeddings.create(input=[edited_text], model=model).data[0].embedding
```

```python
datapath = " " #path for where files are located
```

Following code was used to obtain embedding vectors of ROS

```python
df_ros = pd.read_csv(datapath + "ros.csv")
```

```python
df_ros['Embeddings'] = [None] * len(df_ros)
for index in df_ros.index:
    ros_text = df_ros.loc[index, "Description"]
    df_ros.at[index, "Embeddings"] = get_embedding(ros_text)
```

```python
df_ros.to_csv(datapath + "roswithembeddings.csv", index = False) #save
```

Following code was used to obtain embedding distance of patient visits to ROS 

```python
#read csv file of each visit for patient X
#patients had their own csv file with their visit HPIs
#seprating the path to common folder they were stored in allowed
#for easy change of individual csv file name
df = pd.read_csv(datapath + " ")    
df_ros = pd.read_csv(datapath + "roswithembeddings.csv")    
```

```python
#get embeddings 
df['HPIEmbeddings'] = [None] * len(df)
for index in df.index:
    hpi_text = df.loc[index, "HPI"]
    df.at[index, "HPIEmbeddings"] = get_embedding(hpi_text)
```

```python
#make matrix to store embeddings 
matrix_ros = np.array(df_ros.Embeddings.apply(literal_eval).to_list())
matrix_visits = np.array(df.HPIEmbeddings.apply(literal_eval).to_list())
```

```python
#create array for distance values
distances = np.zeros((len(df_ros),len(df))) #rows = ros, #columns = visit
distances.shape #shape should be 13 x num of visits
```

```python
#calculate Euclidean distances
c = 0
for i in matrix_visits:
    r = 0
    for j in matrix_ros:
        if r < len(df_ros):                     
            distances[r,c] = distance.euclidean(i,j)
        r += 1
    c += 1
```

```python
#save distances into files
df_dist = pd.DataFrame(distances)
df_dist.to_excel(datapath + "")
df_dist.to_csv(datapath + "")
```
